**Requirement:** "Create a function that determines if a year is a leap year. Leap years are divisible by 4, except for years divisible by 100 (which are not leap years), unless they're also divisible by 400 (which are leap years)."

### Step 1: Understand the Requirement

Write in plain language what the function should do:

### Step 2: Apply Systematic Test Design Framework

**Framework 1: Given/When/Then Structure**

This helps you think clearly about test scenarios:
- **Given** (context): What's the starting state?
- **When** (action): What operation happens?
- **Then** (outcome): What should result?

**Framework 2: Equivalence Partitioning**

Group inputs into categories that should behave the same way:
- What are the different "classes" of inputs?
- What's a representative example from each class?

**Apply to Leap Year:**

````python
"""
Equivalence Classes for Leap Year:

1. LEAP: Years divisible by 400
   Example: 2000

2. NOT LEAP: Years divisible by 100 but not 400
   Example: 1900

3. LEAP: Years divisible by 4 but not 100
   Example: 2024

4. NOT LEAP: Years not divisible by 4
   Example: 2023

"""
````

**Now write as Given/When/Then:**

````python
"""
Test Case 1:
Given: A year divisible by 400 (like 2000)
When: I check if it's a leap year
Then: Result should be True

Test Case 2:
Given: A year divisible by 100 but not 400 (like 1900)
When: I check if it's a leap year
Then: Result should be False

Test Case 3:
Given: A year divisible by 4 but not 100 (like 2024)
When: I check if it's a leap year
Then: Result should be True

Test Case 4:
Given: A year not divisible by 4 (like 2023)
When: I check if it's a leap year
Then: Result should be False



[Continue for other equivalence classes...]

"""
````

**Boundary Analysis:**

````python
"""
At the "boundaries" between categories, test both sides:

Boundary at 100 divisibility:
- 1896: leap (before boundary)
- 1900: not leap (at boundary)
- 1904: leap (after boundary)

Boundary at 400 divisibility:
- 1996: leap (before boundary)
- 2000: leap (at boundary - 400 override exceptional)
- 2004: leap (after boundary)

"""
````

### Step 3: Evaluate Test Adequacy

Before writing assertions, answer these strategic questions:

````python
"""
Test Adequacy Check:

1. Have I covered all equivalence classes?
   □ Yes

2. Have I tested boundaries between classes?
   □ Yes

3. What's the RISK if this function fails?
    Medium / High
    (Reasoning: If a time or database function handles leap years incorrectly, it can desynchronize transaction logs, break interest rate calculations, or freeze execution loops on February 29th.)

4. Based on risk, how many tests do I need?
   My decision:
   - Medium-to-High risk: I will implement a combination of equivalence class tests (4 tests) and absolute logical boundary checks (another 4-6 tests) to verify that both the century limit and the 400 override flip smoothly.

5. Am I over-testing? (Could I remove tests without losing confidence?)
No. Because the leap year logic contains multiple nested exceptions (divisible by 4, but not 100, unless 400), each unique case represents a distinct path through the code. Removing any of these test cases would create a blind spot.

"""
````

### Step 4: Write Executable Specifications

Now convert your scenarios to assertions:

````python
def is_leap_year(year):
    """
    Determine if a year is a leap year.

    Rules:
    - Divisible by 4: leap year
    - Divisible by 100: NOT a leap year
    - Divisible by 400: leap year
    """
    # Rule 1: Centennial check divisible by 400 overrides everything
    if year % 400 == 0:
        return True
    # Rule 2: Other  centennial years are not leap years
    if year % 100 == 0:
        return False
    # Rule 3: Standard leap year check
    return year % 4 == 0

# Write your assertions here (one per equivalence class minimum):
# Class 1: Divisible by 400 (Leap Year)
assert is_leap_year(2000) == True, "Failed on Class 1: 2000 should be a Leap Year"
# Class 2: Divisible by 100 but not 400 (Normal Year)
assert is_leap_year(1900) == False, "Failed on Class 2: 1900 should be a Normal Year"
# Class 3: Divisible by 4 but not 100 (Leap Year)
assert is_leap_year(2024) == True, "Failed on Class 3: 2024 should be a Leap Year"
# Class 4: Not divisible by 4 (Normal Year)
assert is_leap_year(2023) == False, "Failed on Class 4: 2023 should be a Normal Year"

# Extra Boundary Analysis Checks (From Step 2 notes)
assert is_leap_year(1896) == True, "Boundary check failed for 1896"
assert is_leap_year(1904) == True, "Boundary check failed for 1904"
assert is_leap_year(1996) == True, "Boundary check failed for 1996"

print("All is_leap_year assertions passed successfully!")
````

**Paste AI's implementation here:**

````python
# AI's implementation:
def is_leap_year(year):
    """
    Determine if a year is a leap year based on standard calender rules.
    Uses basic Python if_statements and boolean logic operators.
    """
    # If a year is divisible by 400, it is automatically a leap year
    if year % 400 == 0:
        return True

    # Otherwise, if it is divisible by 100, but not 400, it is not a leap year
    if year % 100 == 0:
        return False

    # Otherwise, it just needs to be cleanly divisible by 4
    return year % 4 == 0
````

### Step 1: Risk Assessment

### Step 2: Equivalence Partitioning

````python
"""
Identify equivalence classes (categories that behave the same):

VALID class (all requirements met):
- Example: "P@ssword1" (Length 9, has uppercase, lowercase, digit, and special character)

INVALID classes (each requirement violation):
- Too short (< 8 chars): Example: "P@ss1"
- Missing uppercase: Example: "p@ssword1!"
- Missing number: Example: "P@ssword!"
- Missing special char: Example: "Password1"
- Multiple violations: Example: "pass"

Do I need to test EVERY invalid combination?
Decision: No
Reasoning: Testing every combination create exponential redundancy. Testing individual parameter violations ensures coverage of each validation check independently.

"""
````

### Step 3: Boundary Analysis

````python
"""
Key boundaries to test:

Length boundary (8 characters):
- 7 chars (just under): Example: "P@sswo1!"
- 8 chars (exactly at): Example: "P@ssword1!"
- 9 chars (just over): Example: "P@ssword12!"


"""
````

### Step 4: Test Adequacy Decision

````python
"""
How many tests is enough?

Minimum viable: 6 tests (one per equivalence class)
My plan: 9 testsc(equivalence classes + absolute length boundaries)

Why this number?
It targets every unique code branch condition independently while verifying the character-length transition edge points.


Could I get the same confidence with fewer tests?
No. Cutting tests would mean leaving out critical parameter, risking a security blind spot.


What am I NOT testing, and why is that OK?
I am not testing Unicode emojis or special whitespace character. It is OK because the core business requirement specifies standard printable punctuatio characters.


"""
````

### Step 5: Write Specifications

````python


def is_strong_password(password):
    """Validate password strength based on multi-parameter bounds."""
    if len(password) < 8:
        return False

    has_uppercase = any(c.isupper()  for c in password)
    has_lowercase = any(c.islower()  for c in password)
    has_digit = any(c.isdigit() for c in password)
    has_special = any(c in "!@#$%^&*" for c in password)

    return has_uppercase and has_lowercase and has_digit and has_special

# ========================== TEST SPECIFICATIONS ==================================

# Valid password (meets all requirements):
# assert is_strong_password("Pass123!") == True

# Invalid passwords (equivalence class violations):
assert is_strong_password("Pass12!") == False  # too short
assert is_strong_password("pass123!") == False # Missing uppercase
assert is_strong_password("Password!") == False # Missing number
assert is_strong_password("Pass1234") == False # Missing special char

# Boundary tests (if risk justifies):
assert is_strong_password("P@sswor1") == True  # Exactly 8 characters
assert is_strong_password("P@ssword") == False  # Too short

print("All assertions passed successfully!")
````

**Step 2:** Get AI's implementation and paste it below:

````python
# AI's implementation:
def is_strong_password(password):
    """
    Validate a password based on character length and complexity rules.
    Returns True if strong, False otherwise.
    """
    # Rule 1: Minimum length check
    if len(password) < 8:
        return False
    # Standard state check tracking metrics
    has_uppercase = False
    has_lowercase = False
    has_digit = False
    has_special = False

    special_characters = "!@#$%^&*"

    # Evaluate individual characters in sequence
    for char in password:
        if char.isupper():
            has_uppercase = True
        elif char.islower():
            has_lowercase = True
        elif char.isdigit():
            has_digit = True
        elif char in special_characters:
            has_special = True

    # The password is only strong if all conditions are met
    return has_uppercase and has_lowercase and has_digit and has_special
````
