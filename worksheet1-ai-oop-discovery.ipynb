### Task 3.2: Adding Business Logic

Now let's add some business intelligence. Ask AI to help:

> "I want my transaction template to be smarter. It should be able to tell me: 1) If a transaction is a large expense (over $50), 2) How to display it nicely for business reports, and 3) Whether it happened this month. How would you add these business features?"

```python
# Enhanced Transaction class with business logic:
from datetime import datetime

class Transaction:
    def __init__(self, date, description, amount, category):
        self.date = datetime.strptime(date, "%Y-%m-%d")
        self.description = description
        self.amount = amount
        self.category = category

   def is_large_expense(self):
       return self.amount > -50

   def is_this_month(self):
       today = datetime.now()
       return self.date.month == today.month and  self.date.year == today.year

  def display(self):
      return f'{self.date.strftime('%Y-%m-%d')} | {self.description} | ${self.amount:2f} | {self.category}"

```

## Part 4: System-Level Thinking

### Task 4.1: The Manager Class Challenge

Individual transactions are useful, but businesses need to analyse collections of transactions. This is where system-level objects come in.

Ask AI to design a system:

> "I have individual transactions working well, but I need a system to manage hundreds of them for business reporting. I need to calculate total expenses, group spending by category, and find the biggest purchases. How would you design a 'transaction manager' system that organizes all my individual transactions?"

```python
# AI's TransactionManager design:
class TransactionManager:
    def __init__ (self):
        self.transactions = []

    def add_transaction(self,transaction):
        self.transactions.append(transaction)

    def total_expenses(self):
        return sum(t.amount for t in self.transactions if t.category !="Income")

    def total_income(self):
        return sum (t.amount for t in self.transactions if t.category == "Income")

    def group_by_category(self):
        categories = {}

        for t in self.transactions:
            if t.category not in categories:
                categories[t.category] = 0

            categories[t.category] += t.amount

        return categories

    def largest_expense(self):
        expenses = [t for t in self.transactions if t.category != "Income"]

        if len(expenses) == 0:
            return None 

        return min(expenses, key=lambda t: t.amount)
                
    
    
    ```

# Advanced CSV integration:
import csv

class TransactionManager:
    def __init__(self):
        self.transactions= []

    def add_transaction(self, transaction):
        self.transactions.append(transaction)

    def add_manual_transaction(self, date, description, amount, category):

        self.transactions.append(transaction)

    def load_from_csv(self, filename):
        with open(filename, newline="") as file:
            reader = csv.DictReader(file)

            for row in reader:
                transaction = Transaction(
                    row["date"],
                    row["description"],
                    float(row["amount"]),
