# Exercise Set 1: Sales Data Analysis 📊

## Business Context

You're working as a Business Analyst for an office supply company. Your manager has given you sales data and asked for insights to help plan next quarter's inventory and sales strategy.

## Data Overview

We'll use the `sales_data.csv` file with columns:
- **Date**: Sale date
- **Product**: Product name
- **Quantity**: Units sold
- **Unit_Price**: Price per unit (with $ signs)
- **Total_Sale**: Total transaction value
- **Sales_Rep**: Employee who made the sale
- **Region**: Geographic region

````python
import pandas as pd

sales_df = pd.read_csv('/content/sales_data (1).csv')


print(sales_df.head())
print(f"\nDataset: {len(sales_df)} transactions across {sales_df['Region'].nunique()} regions")
````

## Practice Exercise 1.1: Sales Performance by Region

::: {.callout-tip}
## Your AI Collaboration Challenge

**Business Question**: "Which region is performing best, and what's the average sale size by region?"

**Your Task**: Practice prompting AI with this business context. Try creating a prompt that asks for:
1. Total sales by region
2. Average sale amount by region  
3. Number of transactions by region
4. Results formatted for a manager presentation

**Write your prompt below, then see what AI generates:**
:::

**Your Prompt:**
```
Please provide the following sales analysis for each region:
1. Total sales by region
2. Average sale size by region
3. Number of transactions by region
4. Format the results for a business manager presentation
```

**AI Response:**

````python
# AI Response to the Prompt

# Preprocessing: Convert 'Total_Sale' to numeric
# Remove '$' and ',' characters and convert to float
sales_df['Total_Sale'] = sales_df['Total_Sale'].replace({r'\$': '', r',': ''}, regex=True).astype(float)

#Grouping the data by 'Region' and calculating total sales, average sales, and number of transactions
region_summary = sales_df.groupby('Region').agg(
    total_sales=('Total_Sale', 'sum'),
    avg_sales=('Total_Sale', 'mean'),
    transaction_count=('Total_Sale', 'count')
)

#Formatting the results for a business manager presentation
region_summary = region_summary.sort_values(by='total_sales', ascending=False)

#Displaying the formatted results
print("Sales Performance by Region (Formatted for Manager Presentation)")
print(region_summary)
````

**Your Critique:**
- What did AI do well? AI successfully provided the total revenue, total units sold, and profit margins for each region, as requested. The data is sorted by region, which aligns with the goal of identifying the performance of each region. The AI has given a clear numerical output, which is easy  to interpret.
- What would you ask AI to improve? The response lacks a clear formatting for presentation. It could be improved by clearly labeling each column (e.g., as a table or sorted list for easy comparison)
- Does the output help answer the business question?The output answers the question about the regions performing the best, but it does not fully address the business challenge about identifying top products.

## Practice Exercise 1.2: Top Performing Products

::: {.callout-tip}
## Business Scenario

Your inventory manager needs to know which products to stock more of next quarter.

**Challenge**: Ask AI to identify the top 5 products by:
- Total revenue generated
- Total units sold
- Which products have the highest profit margins (hint: compare unit price trends)

**Practice better prompting**: Be specific about formatting and business context.
:::

**Your Prompt:**
`Please analyze the sales data and identify the top 5 products based on the following criteria:
1.Total revenue generated
2. Total units sold
3. Highest profit margins (hint: compare unit price trends)
Format the results clearly, including the product nme, total revenue, total units sold, and profit margins.``

```

**AI Response:**

````python
# AI Response to the Prompt

# Preprocessing: Convert 'Unit_Price' and 'Quantity' to numeric
sales_df['Unit_Price'] = sales_df['Unit_Price'].replace({r'\$': '', r',': ''}, regex=True).astype(float)
sales_df['Quantity'] = sales_df['Quantity'].replace({r'\$': '', r',': ''}, regex=True).astype(int)

# Calculate the total revenue per product
sales_df['Revenue'] = sales_df['Quantity'] * sales_df['Unit_Price']

#Grouping by 'Product' and aggregating the total revenue, total units sold, and profit margins
product_summary = sales_df.groupby('Product').agg(
    total_revenue=('Revenue', 'sum'),
    total_units_sold=('Quantity', 'sum'),
    avg_unit_price=('Unit_Price', 'mean')
)

# Calculate profit margin as (unit price / average unit price) for comparison
product_summary['profit_margin'] = product_summary['avg_unit_price'] / product_summary['avg_unit_price'].max()

# Sorting the products by total revenue, total units sold, and profit margin
top_products = product_summary.sort_values(by=['total_revenue', 'total_units_sold', 'profit_margin'], ascending=False)

# Display the top 5 products fo the inventory manager
top_5_products = top_products.head(5)

# Displaying the formatted results
print("Top 5 Performing Products (Formatted for Manager Presentation)")
print(top_5_products)
````

**Business Insight Questions:**
- Which products should the company stock more of?
- Are high-price items selling well, or is volume more important?
- What would you recommend to the inventory manager?

## Practice Exercise 2.1: Spending Trend Analysis

::: {.callout-tip}
## Business Context

You're helping a friend understand their spending habits over time.

**Challenge**: Ask AI to help you:
1. Convert dates to datetime format
2. Analyze spending trends by day of week
3. Identify if spending has increased over time
4. Spot any unusual spending patterns

**Focus**: Practice asking AI to explain any complex date handling code it provides.
:::

**Your Analysis:**

````python
# 1. Convert the 'Date' column to datetime format
transactions_df['Date'] = pd.to_datetime(transactions_df['Date'])
print(transactions_df.head())

# 2. Analyze spending trends by day of week
transactions_df['Day_of_Week'] = transactions_df['Date'].dt.dayofweek
spending_by_day = transactions_df.groupby('Day_of_Week')['Amount'].sum()

# Map the days to a readable format
spending_by_day.index = spending_by_day.index.map({
    0: 'Monday', 1:'Tuesday', 2:'Wednesday', 3:'Thursday', 4:'Friday', 5:'Saturday', 6:'Sunday'
})

print("Total Spending by Day of Week:")
print(spending_by_day)

# 3. Identify if spending has increased over time (by month)
transactions_df['Month'] = transactions_df['Date'].dt.month
spending_by_month = transactions_df.groupby('Month')['Amount'].sum()

print("Spending by Month:")
print (spending_by_month)

# 4. Spot any unusual spending patterns (using a rolling average)
transactions_df['Rolling_Avg'] = transactions_df['Amount'].rolling(window=7).mean()
print("Spending with Rolling Average:")
print(transactions_df[['Date', 'Amount', 'Rolling_Avg']])
````

## Practice Exercise 2.2: Budget Category Analysis

::: {.callout-tip}
## Personal Finance Scenario

Someone wants to set a realistic monthly budget based on their spending history.

**Challenge**: Use AI to help:
1. Calculate average monthly spending by category
2. Identify categories with high variation (inconsistent spending)
3. Suggest realistic budget amounts for each category
4. Find categories where spending might be reduced

**Advanced**: Ask AI to explain the statistical concepts it uses.
:::

**Your Budget Analysis:**

````python
# 1. Calculate average monthly spending by category
category_avg = transactions_df.groupby('Category')['Amount'].mean()

# 2. Identify categories with high variation (high standard deviation)
category_variance = transactions_df.groupby('Category')['Amount'].std()

# 3. Suggest realistic budget amounts for each category
# A realistic budget could be set at the average amount, but categories with high variability might need adjustments
budget_suggestion = category_avg + (category_variance * 0.5)

# 4. Find categories where spending might be reduced
# Categories with consistently high spending but low variance could be candidates for reduction
reduction_candidates = category_avg[category_variance < category_variance.median()]

# Display results
print("Average Spending by Category:")
print(category_avg)

print("\nCategories with High Variation:")
print(category_variance)

print("\nSuggested Budget Amounts for Each Category:")
print(budget_suggestion)

print("\nCategories That Might Be Reduced:")
print(reduction_candidates)
````

## Scenario 2: Personal Finance Coaching

::: {.callout-note}
## Business Context

You're building a personal finance app feature that provides spending advice.

**Challenge**: Use AI to create code that:
1. Identifies spending categories that are above typical percentages
2. Suggests specific areas for potential savings
3. Provides encouraging feedback for good spending habits
4. Formats advice in a user-friendly way
:::

**Your Coaching Algorithm:**

````python
import pandas as pd

# Example user data: Spending categories (in percentage of income)
user_data = {
    'Category': ['Housing', 'Transportation', 'Entertainment', 'Savings', 'Food'],
    'Spending %': [40, 15, 30, 10, 20]
}

# Typical spending percentages
typical_spending= {
    'Housing': 30,
    'Transportation': 15,
    'Entertainment': 10,
    'Savings': 15,
    'Food': 15
}

#Convert user data to DataFrame
df = pd.DataFrame(user_data)

#Function to give advice on spending
def coaching_algorithm(df, typical_spending):
    advice= []

    for index, row in df.iterrows():
        category = row['Category']
        user_spending = row['Spending %']
        typical_spend= typical_spending[category]

        if user_spending > typical_spend:
          advice.append(f"Your spending on {category} is {user_spending -typical_spend}% above the typical {typical_spend}%. Consider cutting back to save more.")
        elif user_spending < typical_spend:
             advice.append(f"Great job on {category}! You're spending {typical_spend - user_spending}% below the typical {typical_spend}%. Keep it up!")
        else:
             advice.append(f"Your spending on {category} is right on target with the typical {typical_spend}%.Well balanced!")

    return advice

# Get coaching advice
coaching_advice = coaching_algorithm(df, typical_spending)

# Display advice
for line in coaching_advice:
    print(line)
````
