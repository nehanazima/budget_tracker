````python
# 1. Which sales reps are generating the most revenue?
sales_rep_revenue = sales_df.groupby('Sales_Rep')['Total_Sale'].sum().sort_values(ascending=False)

# 2. Who has the highest average sale amount?
sales_rep_avg_sale = sales_df.groupby('Sales_Rep')['Total_Sale'].mean().sort_values(ascending=False)

# 3. Are there any regional patterns in sales rep performance?
regional_sales_rep_performance = sales_df.groupby(['Region', 'Sales_Rep']).agg(
    Total_Revenue=('Total_Sale', 'sum'),
    Average_Sale=('Total_Sale', 'mean'),
    Transaction_Count=('Total_Sale', 'count')
).sort_values(by='Total_Revenue', ascending=False)

# 4. Format the results into a concise and professional summary
print("Sales Performance Review Summary\n" + "=" * 40)

print("\n1. Sales Rep Revenue Performance:")
print("--------------------------------")
print(sales_rep_revenue.map('${:,.2f}'.format).to_string())

print("\n2. Sales Rep Average Sale Amount (Top to Bottom):")
print("----------------------------------------------")
print(sales_rep_avg_sale.map('${:,.2f}'.format).to_string())

print("\n3. Regional Sales Rep Performance:")
print("----------------------------------")
# Format the monetary columns for better readability
regional_sales_rep_performance['Total_Revenue'] = regional_sales_rep_performance['Total_Revenue'].map('${:,.2f}'.format)
regional_sales_rep_performance['Average_Sale'] = regional_sales_rep_performance['Average_Sale'].map('${:,.2f}'.format)
print(regional_sales_rep_performance.to_string())

print("\n\nRecommendations for Review Meeting:\n- Highlight top revenue generators (e.g., " + sales_rep_revenue.index[0] + ") for their contributions.\n- Discuss strategies to improve average sale amounts for reps with lower values.\n- Analyze regional disparities in performance to identify best practices or areas needing support.")
````

## Practice Exercise 2.1: Spending Trend Analysis

::: {.callout-tip}
## Business Context

You're helping a friend understand their spending habits over time.

**Challenge**: Ask AI to help you:
1. Convert dates to datetime format
2. Analyze spending trends by day of week
3. Identify if spending has increased over time
4. Spot any unusual spending patterns

**Focus**: Practice asking AI to explain any complex date handling code it provides.
:::

**Your Analysis:**

````python
# 1. Convert 'Date' to datetime format for easier analysis
# Explanation: The 'Date' column is currently a string. Converting it to datetime objects allows for easy extraction of day of week, month, year, and enables time-series analysis.
transactions_df['Date'] = pd.to_datetime(transactions_df['Date'])
print("Date column converted to datetime format.")

# Extract day of the week for analysis
transactions_df['Day_of_Week'] = transactions_df['Date'].dt.day_name()

# 2. Analyze spending trends by day of the week
# Explanation: Grouping by 'Day_of_Week' and summing 'Amount' (absolute value to count spending) helps identify which days have higher or lower spending. We take the absolute value of Amount because spending is represented as negative numbers.
day_of_week_spending = transactions_df[transactions_df['Amount'] < 0].groupby('Day_of_Week')['Amount'].sum().abs().sort_values(ascending=False)
print("\nSpending by Day of the Week (highest to lowest):")
print(day_of_week_spending.map('${:,.2f}'.format))

# 3. Identify whether the user's spending has increased over time or if it remains consistent.
# Explanation: To see trends over time, we resample the data by month ('ME' for month end) and sum the absolute spending amounts. Plotting this would be ideal, but for a textual summary, we can look at the first and last few months.
monthly_spending = transactions_df[transactions_df['Amount'] < 0].set_index('Date')['Amount'].resample('ME').sum().abs()

print("\nMonthly Spending Trend (first 3 months and last 3 months, if available):")
if len(monthly_spending) > 6:
    print(monthly_spending.head(3).map('${:,.2f}'.format))
    print("...")
    print(monthly_spending.tail(3).map('${:,.2f}'.format))
else:
    print(monthly_spending.map('${:,.2f}'.format))

# Simple check for overall trend (can be improved with regression)
if monthly_spending.diff().mean() > 0:
    print("\nObservation: Spending appears to have generally increased over time.")
elif monthly_spending.diff().mean() < 0:
    print("\nObservation: Spending appears to have generally decreased over time.")
else:
    print("\nObservation: Spending appears to have remained relatively consistent over time.")

# 4. Spot any unusual spending patterns or significant changes in spending behavior.
# Explanation: Outliers in spending can be identified by looking at individual transactions that are significantly larger than the average. We can filter for large individual negative amounts.
mean_spending = transactions_df[transactions_df['Amount'] < 0]['Amount'].abs().mean()
std_spending = transactions_df[transactions_df['Amount'] < 0]['Amount'].abs().std()

# Define a threshold for unusual spending (e.g., 2 standard deviations above the mean)
unusual_spending_threshold = mean_spending + (2 * std_spending)
unusual_transactions = transactions_df[(transactions_df['Amount'] < 0) & (transactions_df['Amount'].abs() > unusual_spending_threshold)].sort_values(by='Amount')

print("\nUnusual Spending Patterns (transactions significantly larger than average):")
if not unusual_transactions.empty:
    # Create a temporary DataFrame to format only the 'Amount' column for printing
    display_unusual_transactions = unusual_transactions[['Date', 'Description', 'Category', 'Amount']].copy()
    display_unusual_transactions['Amount'] = display_unusual_transactions['Amount'].map('${:,.2f}'.format)
    print(display_unusual_transactions.to_string(index=False))
else:
    print("No significantly unusual spending transactions identified based on the 2-sigma rule.")
````

### Scenario 1: Retail Store Analysis - Business Insights

**Assumption**: For this analysis, we are interpreting the `Region` column in `sales_df` as `Store Location`.

````python
# 1. Which stores need more inventory?
# We'll identify stores with the highest total sales revenue and total units sold as potential candidates for increased inventory.

store_performance = sales_df.groupby('Region').agg(
    Total_Revenue=pd.NamedAgg(column='Total_Sale', aggfunc='sum'),
    Total_Units_Sold=pd.NamedAgg(column='Quantity', aggfunc='sum')
).sort_values(by='Total_Revenue', ascending=False).reset_index()

print("\n--- Store Performance (Potential Inventory Needs) ---")
print("-------------------------------------------------------")

# Create a temporary DataFrame for display purposes to format the 'Total_Revenue' column
display_store_performance = store_performance.rename(columns={'Region': 'Store Location'}).copy()
display_store_performance['Total_Revenue'] = display_store_performance['Total_Revenue'].map('${:,.2f}'.format)
print(display_store_performance.to_string(index=False))

print("\n**Insight**: Stores with higher total revenue and units sold (e.g., NSW, VIC) likely have higher demand and may require more inventory to prevent stockouts and maximize sales.")
````

````python
# 2. What products to discontinue?
# We'll identify products with the lowest total revenue and total units sold as potential candidates for discontinuation.

product_performance = sales_df.groupby('Product').agg(
    Total_Revenue=pd.NamedAgg(column='Total_Sale', aggfunc='sum'),
    Total_Units_Sold=pd.NamedAgg(column='Quantity', aggfunc='sum')
).sort_values(by='Total_Revenue', ascending=True).reset_index()

print("\n--- Product Performance (Potential Discontinuation Candidates) ---")
print("--------------------------------------------------------------------")

# Create a temporary DataFrame for display purposes to format the 'Total_Revenue' column
display_product_performance = product_performance.copy()
display_product_performance['Total_Revenue'] = display_product_performance['Total_Revenue'].map('${:,.2f}'.format)
print(display_product_performance.to_string(index=False))

print("\n**Insight**: Products with very low total revenue and units sold might be underperforming and could be considered for discontinuation to free up shelf space and capital for better-selling items.")
````

````python
# 3. When to schedule promotional campaigns?
# For this, we'd typically analyze sales trends over a longer period (e.g., monthly, quarterly, seasonal).
# Given our current `sales_df` only covers a short period (August 1st to August 10th), a robust seasonal analysis isn't possible.
# However, we can demonstrate how you would approach this by looking at daily trends within the available data.

# Ensure 'Date' column is datetime type (already done in previous steps, but good practice to ensure)
# sales_df['Date'] = pd.to_datetime(sales_df['Date'])

daily_sales = sales_df.groupby('Date')['Total_Sale'].sum().reset_index()

print("\n--- Daily Sales Trends (for Promotional Campaign Strategy) ---")
print("--------------------------------------------------------------")

# Create a temporary DataFrame for display purposes to format the 'Total_Sale' column
display_daily_sales = daily_sales.copy()
display_daily_sales['Total_Sale'] = display_daily_sales['Total_Sale'].map('${:,.2f}'.format)
print(display_daily_sales.to_string(index=False))

print("\n**Insight**: With limited data, we can only observe short-term daily fluctuations. For effective promotional campaign scheduling, a longer sales history is crucial to identify weekly, monthly, and seasonal patterns (e.g., higher sales on weekends, before holidays, etc.).")
````

````python
# 4. How to allocate sales staff?
# We can look at sales representative performance per 'Store Location' (Region) to inform staff allocation.
# This helps identify high-performing reps, or stores that might need more support or staffing.

sales_rep_allocation_analysis = sales_df.groupby(['Region', 'Sales_Rep']).agg(
    Total_Revenue=pd.NamedAgg(column='Total_Sale', aggfunc='sum'),
    Average_Sale_Value=pd.NamedAgg(column='Total_Sale', aggfunc='mean'),
    Number_of_Transactions=pd.NamedAgg(column='Total_Sale', aggfunc='count')
).sort_values(by='Total_Revenue', ascending=False).reset_index()

print("\n--- Sales Staff Allocation Analysis (by Store Location and Rep) ---")
print("-------------------------------------------------------------------")

# Format monetary values for better readability
sales_rep_allocation_analysis['Total_Revenue'] = sales_rep_allocation_analysis['Total_Revenue'].map('${:,.2f}'.format)
sales_rep_allocation_analysis['Average_Sale_Value'] = sales_rep_allocation_analysis['Average_Sale_Value'].map('${:,.2f}'.format)

print(sales_rep_allocation_analysis.rename(columns={'Region': 'Store Location'}).to_string(index=False))

print("\n**Insight**: This data can help allocate staff more effectively. For example, high-performing reps might be assigned to busier stores or used to mentor others. Stores with lower overall revenue might benefit from additional staffing or targeted training for existing staff. Analyzing the number of transactions can also indicate workload.")
````

*   **Product Portfolio**: Consider evaluating or discontinuing products with consistently low sales.
*   **Promotional Strategy**: While detailed campaign timing requires more historical data, this framework helps identify trends to inform future planning.
